```python
import os
import pandas as pd

DRIVE_ROOT = "/content/drive/MyDrive/Amazon ML Challenge 2026"

DATASET_ROOT = os.path.join(
    DRIVE_ROOT,
    "01_Dataset",
    "6ab10eb3b23ba_student_resource",
    "student_resource",
    "dataset"
)

TRAIN_ROOT = os.path.join(DATASET_ROOT, "train")
TEST_ROOT = os.path.join(DATASET_ROOT, "test")

print("Train path:", TRAIN_ROOT)
print("Test path:", TEST_ROOT)

print("\nTrain files:")
print(os.listdir(TRAIN_ROOT))

print("\nTest files:")
print(os.listdir(TEST_ROOT))
```

```python
train_s1 = pd.read_csv(
    os.path.join(TRAIN_ROOT, "train_source1.tsv"),
    sep="\t"
)

train_s2 = pd.read_csv(
    os.path.join(TRAIN_ROOT, "train_source2.tsv"),
    sep="\t"
)

train_s3 = pd.read_csv(
    os.path.join(TRAIN_ROOT, "train_source3.tsv"),
    sep="\t"
)

ground_truth = pd.read_csv(
    os.path.join(TRAIN_ROOT, "train_ground_truth.tsv"),
    sep="\t"
)

test_s1 = pd.read_csv(
    os.path.join(TEST_ROOT, "test_source1.tsv"),
    sep="\t"
)

test_s2 = pd.read_csv(
    os.path.join(TEST_ROOT, "test_source2.tsv"),
    sep="\t"
)

test_s3 = pd.read_csv(
    os.path.join(TEST_ROOT, "test_source3.tsv"),
    sep="\t"
)

print("Loaded successfully!")
```

```python
ground_truth["matched_count"] = (
    ground_truth["matched_entity_ids"]
    .fillna("")
    .astype(str)
    .apply(
        lambda x: 0 if x.strip() == "" else len(x.split(","))
    )
)

print(
    ground_truth["matched_count"]
    .value_counts()
    .sort_index()
)
```

```python
display(ground_truth.head(20))
```

```python
match_counts = ground_truth["matched_count"].value_counts().sort_index()

print(
    pd.DataFrame({
        "count": match_counts,
        "percentage": (match_counts / len(ground_truth) * 100).round(2)
    })
)
```

```python
print("Total S1:", len(ground_truth))
print("Singletons:", (ground_truth["matched_count"] == 0).sum())
print(
    "Singleton %:",
    round((ground_truth["matched_count"] == 0).mean() * 100, 2)
)
```

```python
matches = ground_truth["matched_entity_ids"].fillna("").astype(str)

s2_count = matches.str.count(r"S2-")
s3_count = matches.str.count(r"S3-")

print("Total S2 matches:", s2_count.sum())
print("Total S3 matches:", s3_count.sum())

print("\nNumber of S1 entities with:")
print("S2 matches only :", ((s2_count > 0) & (s3_count == 0)).sum())
print("S3 matches only :", ((s2_count == 0) & (s3_count > 0)).sum())
print("Both S2 and S3  :", ((s2_count > 0) & (s3_count > 0)).sum())
print("No matches      :", ((s2_count == 0) & (s3_count == 0)).sum())
```

```python
s1_lookup = train_s1.set_index("entity_id")
s2_lookup = train_s2.set_index("entity_id")
s3_lookup = train_s3.set_index("entity_id")

def show_match(s1_id, matched_ids):
    print("\nS1")
    print(s1_lookup.loc[s1_id][
        ["business_name", "business_address", "country"]
    ])

    for mid in matched_ids.split(","):
        if mid.startswith("S2-"):
            row = s2_lookup.loc[mid]
        else:
            row = s3_lookup.loc[mid]

        print("\n", mid)
        print(row[
            ["business_name", "business_address", "country"]
        ])
```

```python
for _, row in ground_truth[
    ground_truth["matched_count"] > 0
].sample(10, random_state=42).iterrows():

    show_match(
        row["source1_entity_id"],
        row["matched_entity_ids"]
    )
```

```python
train_countries = set(
    pd.concat([
        train_s1["country"],
        train_s2["country"],
        train_s3["country"]
    ]).dropna()
)

test_countries = set(
    pd.concat([
        test_s1["country"],
        test_s2["country"],
        test_s3["country"]
    ]).dropna()
)

print("Train countries:", train_countries)
print("Test countries:", test_countries)
print("New test countries:", test_countries - train_countries)
```

```python
# Build lookups once
s1_lookup = train_s1.set_index("entity_id")
s2_lookup = train_s2.set_index("entity_id")
s3_lookup = train_s3.set_index("entity_id")


# Get only S1 entities that have matches
matched_gt = ground_truth[
    ground_truth["matched_entity_ids"].fillna("").str.strip() != ""
]

print("S1 entities with at least one match:", len(matched_gt))


# Show 20 examples
sample = matched_gt.sample(
    min(20, len(matched_gt)),
    random_state=42
)

for _, row in sample.iterrows():

    s1_id = row["source1_entity_id"]

    print("\n" + "=" * 80)

    print("S1:", s1_id)
    print("Name   :", s1_lookup.loc[s1_id, "business_name"])
    print("Address:", s1_lookup.loc[s1_id, "business_address"])
    print("Country:", s1_lookup.loc[s1_id, "country"])

    matched_ids = row["matched_entity_ids"].split(",")

    for mid in matched_ids:

        mid = mid.strip()

        if mid.startswith("S2-"):
            lookup = s2_lookup
        elif mid.startswith("S3-"):
            lookup = s3_lookup
        else:
            continue

        print("\n", mid)
        print("Name   :", lookup.loc[mid, "business_name"])
        print("Address:", lookup.loc[mid, "business_address"])
        print("Country:", lookup.loc[mid, "country"])
```
